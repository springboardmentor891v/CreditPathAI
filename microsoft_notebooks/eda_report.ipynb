```python
import pandas as pd
import os

# Define the text files to combine
text_files = ['Loan.txt', 'loan_prod.txt', 'borrower.txt', 'Borrower_Prod.txt']

def combine_txt_files_to_csv():
    """
    Combine multiple tab-separated text files into a single CSV file
    """
    all_dataframes = []
    
    print("📁 Processing text files...")
    
    for file in text_files:
        if os.path.exists(file):
            try:
                # Read tab-separated file
                df = pd.read_csv(file, sep='\t')
                
                # Add source file identifier
                df['source_file'] = file
                
                all_dataframes.append(df)
                print(f"✅ Loaded {file}: {df.shape[0]} rows, {df.shape[1]} columns")
                print(f"   Columns: {list(df.columns[:5])}{'...' if len(df.columns) > 5 else ''}")
                
            except Exception as e:
                print(f"❌ Error loading {file}: {e}")
        else:
            print(f"⚠️ File not found: {file}")
    
    if all_dataframes:
        # Combine all DataFrames
        print("\n🔄 Combining all data...")
        combined_df = pd.concat(all_dataframes, ignore_index=True, sort=False)
        
        # Remove the source_file column if you don't need it
        # combined_df = combined_df.drop('source_file', axis=1)
        
        # Save to CSV
        output_file = 'combined_loan_dataset.csv'
        combined_df.to_csv(output_file, index=False)
        
        print(f"🎉 Successfully created {output_file}")
        print(f"📊 Final dataset shape: {combined_df.shape}")
        print(f"📋 Total columns: {combined_df.shape[1]}")
        
        # Show summary
        print("\n📈 Dataset Summary:")
        print(f"   - Total records: {len(combined_df):,}")
        print(f"   - Total features: {combined_df.shape[1]}")
        print(f"   - Missing values: {combined_df.isnull().sum().sum()}")
        
        # Show first few rows
        print("\n👀 First 5 rows:")
        display(combined_df.head())
        
        return combined_df
    else:
        print("❌ No files were successfully loaded")
        return None

# Execute the function
combined_data = combine_txt_files_to_csv()
```
