```python
import pandas as pd

df = pd.read_csv('/content/Loan_Default.csv')

display(df.head())
display(df.info())
```

```python
missing_values = df.isnull().sum()
missing_values = missing_values[missing_values > 0]

numeric_cols_with_missing = df[missing_values.index].select_dtypes(include=['float64', 'int64']).columns
categorical_cols_with_missing = df[missing_values.index].select_dtypes(include=['object']).columns

for col in numeric_cols_with_missing:
    df[col].fillna(df[col].median(), inplace=True)

for col in categorical_cols_with_missing:
    df[col].fillna(df[col].mode()[0], inplace=True)

display(df.isnull().sum().sum())
```

```python
import numpy as np

numeric_cols = df.select_dtypes(include=np.number).columns

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df[col] = df[col].apply(lambda x: upper_bound if x > upper_bound else (lower_bound if x < lower_bound else x))

display(df.describe())
```

```python
categorical_cols = df.select_dtypes(include='object').columns

threshold = 0.05

for col in categorical_cols:
    value_counts = df[col].value_counts(normalize=True)
    infrequent_categories = value_counts[value_counts < threshold].index
    df[col] = df[col].replace(infrequent_categories, 'Other')
    print(f"Value counts for {col} after grouping infrequent categories:")
    display(df[col].value_counts())
```

```python
categorical_cols = df.select_dtypes(include='object').columns

df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

display(df.head())
display(df.info())
```

```python
from sklearn.preprocessing import MinMaxScaler

numeric_cols = df.select_dtypes(include=['float64', 'int64']).columns.tolist()
if 'ID' in numeric_cols:
    numeric_cols.remove('ID')
if 'Status' in numeric_cols:
    numeric_cols.remove('Status')

scaler = MinMaxScaler()
df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

display(df.head())
```

```python
import numpy as np
from sklearn.preprocessing import PolynomialFeatures

if 'rate_of_interest' in df.columns and 'Interest_rate_spread' in df.columns:
    df['interest_rate_x_spread'] = df['rate_of_interest'] * df['Interest_rate_spread']
else:
    print("Warning: 'rate_of_interest' or 'Interest_rate_spread' column not found. 'interest_rate_x_spread' not created.")

if 'loan_amount' in df.columns and 'income' in df.columns and (df['income'] != 0).any():
    df['loan_to_income'] = df['loan_amount'] / df['income']
    max_loan_to_income = df['loan_to_income'][np.isfinite(df['loan_to_income'])].max()
    if np.isinf(max_loan_to_income):
         df['loan_to_income'].replace([np.inf, -np.inf], 0, inplace=True)
        df['loan_to_income'].replace([np.inf, -np.inf], max_loan_to_income, inplace=True)

    df['loan_to_income'].fillna(df['loan_to_income'].median(), inplace=True)
else:
    df['loan_to_income'] = 0
    print("Warning: 'loan_amount' or 'income' column not found or 'income' contains only zero values. 'loan_to_income' created with placeholder values.")


if 'Credit_Score' in df.columns:
    poly = PolynomialFeatures(degree=2, include_bias=False)
    credit_score_poly = poly.fit_transform(df[['Credit_Score']])
    poly_feature_names = [f'Credit_Score_poly_{i}' for i in range(credit_score_poly.shape[1])]
    credit_score_poly_df = pd.DataFrame(credit_score_poly, columns=poly_feature_names, index=df.index)
    df = pd.concat([df, credit_score_poly_df], axis=1)
else:
    print("Warning: 'Credit_Score' column not found. Polynomial feature for 'Credit_Score' not created.")


display(df.head())
```

```python
continuous_cols = ['loan_amount', 'rate_of_interest', 'Interest_rate_spread', 'Upfront_charges', 'property_value', 'income', 'LTV', 'dtir1']

for col in continuous_cols:
    if col in df.columns:
        df[f'{col}_binned'] = pd.qcut(df[col], q=10, labels=False, duplicates='drop')
        df = df.drop(columns=[col])
    else:
        print(f"Column '{col}' not found in the DataFrame.")


display(df.head())
if 'loan_amount_binned' in df.columns:
    display(df['loan_amount_binned'].value_counts().sort_index())
elif 'rate_of_interest_binned' in df.columns:
     display(df['rate_of_interest_binned'].value_counts().sort_index())
```

```python
import numpy as np
from scipy.stats import skew, boxcox

numeric_cols = df.select_dtypes(include=np.number).columns

skewness = df[numeric_cols].apply(lambda x: skew(x.dropna()))
skewed_cols = skewness[abs(skewness) > 0.5].index

for col in skewed_cols:
    if (df[col] > 0).all():
        if abs(skewness[col]) > 1:
             df[col], _ = boxcox(df[col])
        else:
            df[col] = np.log1p(df[col])
    else:
        df[col] = np.sqrt(df[col] + abs(df[col].min()))

transformed_skewness = df[skewed_cols].apply(lambda x: skew(x.dropna()))

print("Original Skewness:")
display(skewness[skewed_cols])
print("\nTransformed Skewness:")
display(transformed_skewness)
```

```python
from collections import Counter
from imblearn.over_sampling import SMOTE
from sklearn.model_selection import train_test_split

target_variable = 'Status'

print("Original dataset shape %s" % Counter(df[target_variable]))

X = df.drop(target_variable, axis=1)
y = df[target_variable]

if len(y.unique()) > 1:
    smote = SMOTE(random_state=42)
    X_res, y_res = smote.fit_resample(X, y)
    print('Resampled dataset shape %s' % Counter(y_res))
else:
    print("The target variable has only one class. SMOTE cannot be applied.")
```
