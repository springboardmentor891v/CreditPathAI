```python


import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


sns.set_style('whitegrid')
```

looking at the data first

```python

print("Data Shape:")
print(df.shape)
print("\nData Info:")
df.info()
```

```python

df.head()
```

 cleaning up the data

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
df.columns = df.columns.str.lower().str.replace(' ', '_')
```

```python

df.drop(columns=['upfront_charges', 'interest_rate_spread'], inplace=True)
df.head()
```

```python
df = df.drop(columns=['id', 'year'])
```

```python
df.columns
```

```python
df['has_co_applicant'] = df['co-applicant_credit_type'].notna().astype(int)
df = df.drop(columns=['co-applicant_credit_type'])
```

```python
df.columns
```

```python
df.head()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='open_credit', data=df)
plt.title('Count of Open Credit')
plt.xlabel('Open Credit')
plt.ylabel('Count')
plt.show()
```

```python
columns_to_check = [
    'open_credit', 'loan_limit', 'approv_in_adv', 'credit_worthiness',
    'business_or_commercial', 'neg_ammortization', 'interest_only',
    'lump_sum_payment', 'construction_type', 'secured_by',
    'submission_of_application', 'security_type'
]

for col in columns_to_check:
    print(f"Value counts for '{col}':")
    print(df[col].value_counts())
    print("\n" + "="*30 + "\n")
```

```python
# In pandas:
df = df.drop(columns=['construction_type', 'secured_by', 'security_type'])
```

```python
df.columns
```

```python
df.drop(columns=['open_credit'], inplace=True)
```

```python
df.columns
```

```python
df['status'].value_counts()
```

```python
plt.figure(figsize=(8, 6))
sns.countplot(x='status', data=df)
plt.title('Distribution of Loan Status')
plt.xlabel('Loan Status (0: Non-Default, 1: Default)')
plt.ylabel('Count')
plt.xticks([0, 1], ['Non-Default', 'Default'])
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='status', y='loan_amount', data=df)
plt.title('Loan Amount vs. Loan Status')
plt.xlabel('Loan Status (0: Non-Default, 1: Default)')
plt.ylabel('Loan Amount')
plt.xticks([0, 1], ['Non-Default', 'Default'])
plt.show()
```
